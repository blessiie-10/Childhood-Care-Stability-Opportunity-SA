This notebook checks the downloaded person and household files before building the child-level analysis dataset.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(ROOT / 'src'))
ROOT
```

```python
from inspect_ghs import discover_files, read_data

person_path, household_path = discover_files()
person_path, household_path
```

```python
person = read_data(person_path)
household = read_data(household_path)
person.columns = person.columns.str.strip().str.lower()
household.columns = household.columns.str.strip().str.lower()

print('Person shape:', person.shape)
print('Household shape:', household.shape)
```

```python
parent_columns = ['age', 'hhc_moth_parthh', 'hhc_fath_parthh', 'person_wgt']
person[parent_columns].head(10)
```

```python
for column in ['hhc_moth_parthh', 'hhc_fath_parthh', 'edu_attend', 'edu_same']:
    print(f'\n{column}')
    print(person[column].value_counts(dropna=False).sort_index())
```
